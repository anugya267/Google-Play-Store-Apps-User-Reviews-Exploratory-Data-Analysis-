## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.
#creating a copy of the current dataset
df1=data.copy()
```

```python
#analysing the null value
df1.isnull().sum()
```

```python
#checking the data types
df1.info()
```

#Filling the missing values

```python
#check the skewness of 'Rating' column to decide correct imputation method
skew=df1['Rating'].skew()
print(skew) #negative skewness indicates a left skewed distribution
```

```python
#filling missing value with median
rat_med=df1['Rating'].median()
df1['Rating'].fillna(rat_med,inplace=True)
```

```python
#droping rows with missing values in reviews/sentiment-related column
#keeping them would bias so we ensure data quality by removing incomplete records
df1.dropna(subset=['Translated_Review','Sentiment','Sentiment_Polarity','Sentiment_Subjectivity'],inplace=True)
```

```python
#drop duplicate rows if any
df1.drop_duplicates(inplace=True)
#reset index after cleaning
df1.reset_index(drop=True,inplace=True)
```

###Imputation is not used for review and sentiment columns-
Imputation was not used for review and sentiment columns because they contain textual data and sentiment scores derived directly from the text. Imputing these values would require guessing user opinions, which could introduce bias and distort the analysis.

```python
#check duplicate values
df1.duplicated().sum()
```

```python
#converting wrong datatypes
#'Installs' column is stored as object(string) with '+' and ','
df1['Installs'] = df1['Installs'].astype(str)
df1['Installs'] = df1['Installs'].str.replace('[+,]', '', regex=True)
df1['Installs'] = df1['Installs'].astype(int)
```

```python
#'Review' column is stored as object but it is stored as number
df1['Reviews'] = df1['Reviews'].astype(int)
```

```python
#convert the price column from string (due to '$')to float and replace empty string with nan
df1['Price'] = df1['Price'].astype(str)
df1['Price'] = df1['Price'].str.replace(r'[^0-9]', '', regex=True)
df1['Price']=df1['Price'].replace('',np.nan)
df1['Price'] = df1['Price'].astype(float)
```

```python
#drop rows where size is 'varies with device' or missing
df1=df1[df1['Size'].notna()&(df1['Size']!='Varies with device')].copy()
#convert remaining size value to float
# Convert remaining size values to float (in MB)
def convert_size(size):
    size = str(size).strip()

    if 'M' in size:
        return float(size.replace('M', ''))
    elif 'k' in size or 'K' in size:
        return float(size.replace('k', '').replace('K', '')) / 1024
    else:
        return float(size)

# Apply function to Size column
df1['Size'] = df1['Size'].apply(convert_size)
```

```python
#convert 'Last Updated' column from object type to datetime format
df1['Last Updated'] = pd.to_datetime(df1['Last Updated'])
```

```python
#convert Sentiment column to float
df1['Sentiment_Polarity'] = df1['Sentiment_Polarity'].astype(float)
df1['Sentiment_Subjectivity'] = df1['Sentiment_Subjectivity'].astype(float)
```

```python
#Extract Year and Month from Last Updated
df1['Year'] = df1['Last Updated'].dt.year
df1['Month'] = df1['Last Updated'].dt.month
```
